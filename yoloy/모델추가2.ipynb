```python
from shapely.geometry import LineString

WIDTH_PX = 12       # 균열 선 두께(원본 픽셀). 8 / 12 / 16 등으로 바꿔 비교해 볼 것
EXTRA_CRACK = 0     # train에 추가할 균열 이미지 수 (처음엔 0, 나중에 1500 등으로 늘린다)
TAG = f"w{WIDTH_PX}_e{EXTRA_CRACK}"   # 설정이 다르면 폴더·실행 이름도 달라진다
V2_DIR = BASE_DIR / "조민서" / f"yolo_seg_v2_{TAG}"
LOCAL_V2 = Path(f"/content/yolo_seg_v2_{TAG}")
CLASS_MAP = {"균열": 0, "포트홀": 1}

def crack_polygon(flat, w):
    pts = list(zip(flat[0::2], flat[1::2]))
    if len(pts) < 2: return None
    ls = LineString(pts)
    if ls.length == 0: return None
    poly = ls.buffer(w / 2, cap_style=1, join_style=1).simplify(1.0)
    if poly.is_empty: return None
    if poly.geom_type == "MultiPolygon":
        poly = max(poly.geoms, key=lambda g: g.area)
    return list(poly.exterior.coords)[:-1]

def pothole_polygon(flat):
    if len(flat) == 4:
        x, y, w, h = flat
        return [(x, y), (x + w, y), (x + w, y + h), (x, y + h)]
    if len(flat) >= 6:
        return list(zip(flat[0::2], flat[1::2]))
    return None

def poly_area(pts):
    return 0.5 * abs(sum(pts[i][0] * pts[(i + 1) % len(pts)][1] - pts[(i + 1) % len(pts)][0] * pts[i][1] for i in range(len(pts))))

def build_split(files, split, z):
    (V2_DIR / "images" / split).mkdir(parents=True, exist_ok=True)
    (V2_DIR / "labels" / split).mkdir(parents=True, exist_ok=True)
    stats = collections.Counter()
    for fn in files:
        label_out = V2_DIR / "labels" / split / (Path(fn).stem + ".txt")
        img_out = V2_DIR / "images" / split / fn
        if label_out.exists() and img_out.exists():
            stats["skip"] += 1; continue
        d = json.load(z.open(filename_to_zipjson[fn]))
        iw, ih = d["images"][0]["width"], d["images"][0]["height"]
        cat = {c["id"]: c["name"] for c in d["categories"]}
        lines = []
        for a in d["annotations"]:
            name = cat.get(a["category_id"], "")
            cls = "균열" if "균열" in name else ("포트홀" if "홀" in name else None)
            if cls is None: continue
            for s in a["segmentation"]:
                pts = crack_polygon(s, WIDTH_PX) if cls == "균열" else pothole_polygon(s)
                if not pts or len(pts) < 3: stats["bad"] += 1; continue
                if poly_area(pts) > 0.5 * iw * ih: stats["too_big"] += 1; continue
                norm = []
                for x, y in pts:
                    norm += [min(max(x / iw, 0), 1), min(max(y / ih, 0), 1)]
                lines.append(f"{CLASS_MAP[cls]} " + " ".join(f"{c:.6f}" for c in norm))
        src = photo_path_map.get(fn)
        if src is None: stats["no_img"] += 1; continue
        label_out.write_text("\n".join(lines))
        shutil.copy(src, img_out)
        stats["ok"] += 1
    return stats

random.seed(42)
pool = [f for f in crack_map if f not in test_names and f not in set(old_train) and f not in set(old_val)]
extra = random.sample(pool, min(EXTRA_CRACK, len(pool)))
new_train = old_train + extra
print("새 train", len(new_train), "(기존", len(old_train), "+ 추가 균열", len(extra), ") / val", len(old_val))

with zipfile.ZipFile(TL01_ZIP) as z:
    print("train:", dict(build_split(new_train, "train", z)))
    print("val:", dict(build_split(old_val, "val", z)))

(V2_DIR / "data.yaml").write_text(f"path: {V2_DIR}\ntrain: images/train\nval: images/val\n\nnames:\n  0: 도로균열\n  1: 도로홀\n", encoding="utf-8")
```

```python
import time
from ultralytics import YOLO

def evaluate(weights, imgsz=960, tag=""):
    t0 = time.time()
    print(f"[1/3] 모델 불러오는 중... ({tag})")
    m = YOLO(str(weights))
    print(f"[2/3] 검증 사진 544장 채점 중... (아래 진행 막대 확인)")
    r = m.val(data=DATA, imgsz=imgsz, batch=4, plots=False, verbose=True)
    print("[3/3] 점수 정리 중...")
    names = {0: "균열", 1: "포트홀"}
    out = {"box_mAP50": round(float(r.box.map50), 3), "mask_mAP50": round(float(r.seg.map50), 3)}
    for i, c in enumerate(r.box.ap_class_index):
        out[f"box_{names[int(c)]}"] = round(float(r.box.ap50[i]), 3)
        out[f"mask_{names[int(c)]}"] = round(float(r.seg.ap50[i]), 3)
    print(f"\n✅ 완료 ({time.time() - t0:.0f}초 걸림)")
    print(tag, out)
    return out

old_best = RESULT_DIR / "road_damage_detector_seg" / "weights" / "best.pt"
print("4차 모델 파일 있음?", old_best.exists())
baseline = evaluate(old_best, 960, "4차(기존 모델, 새 라벨 val)")
```

```python
# ===== 0. GPU 확인 =====
import torch
assert torch.cuda.is_available(), "❌ GPU가 없습니다. 런타임 유형을 T4 GPU로 바꾼 뒤 다시 실행하세요."
print("✅ GPU:", torch.cuda.get_device_name(0))

# ===== A. 연결·설정 =====
from google.colab import drive
drive.mount('/content/drive')
import json, zipfile, random, shutil, collections, subprocess, gc
from pathlib import Path

BASE_DIR = Path("/content/drive/MyDrive/3-여름학기/미래학술제")
CHECKPOINT_DIR = BASE_DIR / "조민서" / "seg_체크포인트"
RESULT_DIR = BASE_DIR / "조민서" / "yolo_학습결과"
TL01_ZIP = BASE_DIR / "라벨데이터" / "TL01.zip"

crack_map = json.load(open(CHECKPOINT_DIR / "crack_map.json"))
pothole_map = json.load(open(CHECKPOINT_DIR / "pothole_map.json"))
sampling = json.load(open(CHECKPOINT_DIR / "sampling.json"))
photo_path_map = {k: Path(v) for k, v in json.load(open(CHECKPOINT_DIR / "photo_path_map.json")).items()}
filename_to_zipjson = {**crack_map, **pothole_map}
old_train, old_val = sampling["train_files"], sampling["val_files"]
test_names = {p.name for p in (BASE_DIR / "최종테스트셋" / "최종_원천").glob("*.jpg")}
OLD_PREFIX = "/content/drive/MyDrive/미래학술제"
photo_path_map = {k: Path(str(v).replace(OLD_PREFIX, str(BASE_DIR), 1)) for k, v in photo_path_map.items()}

# ===== B. 설치 =====
subprocess.run("pip install -q ultralytics", shell=True)

# ===== C. 라벨 만들기 (균열 1500장 추가, 이미 만든 건 건너뜀) =====
from shapely.geometry import LineString
WIDTH_PX = 12
EXTRA_CRACK = 1500
TAG = f"w{WIDTH_PX}_e{EXTRA_CRACK}"
V2_DIR = BASE_DIR / "조민서" / f"yolo_seg_v2_{TAG}"
LOCAL_V2 = Path(f"/content/yolo_seg_v2_{TAG}")
CLASS_MAP = {"균열": 0, "포트홀": 1}

def crack_polygon(flat, w):
    pts = list(zip(flat[0::2], flat[1::2]))
    if len(pts) < 2: return None
    ls = LineString(pts)
    if ls.length == 0: return None
    poly = ls.buffer(w / 2, cap_style=1, join_style=1).simplify(1.0)
    if poly.is_empty: return None
    if poly.geom_type == "MultiPolygon":
        poly = max(poly.geoms, key=lambda g: g.area)
    return list(poly.exterior.coords)[:-1]

def pothole_polygon(flat):
    if len(flat) == 4:
        x, y, w, h = flat
        return [(x, y), (x + w, y), (x + w, y + h), (x, y + h)]
    if len(flat) >= 6:
        return list(zip(flat[0::2], flat[1::2]))
    return None

def poly_area(pts):
    return 0.5 * abs(sum(pts[i][0] * pts[(i + 1) % len(pts)][1] - pts[(i + 1) % len(pts)][0] * pts[i][1] for i in range(len(pts))))

def build_split(files, split, z):
    (V2_DIR / "images" / split).mkdir(parents=True, exist_ok=True)
    (V2_DIR / "labels" / split).mkdir(parents=True, exist_ok=True)
    stats = collections.Counter()
    for fn in files:
        label_out = V2_DIR / "labels" / split / (Path(fn).stem + ".txt")
        img_out = V2_DIR / "images" / split / fn
        if label_out.exists() and img_out.exists():
            stats["skip"] += 1; continue
        d = json.load(z.open(filename_to_zipjson[fn]))
        iw, ih = d["images"][0]["width"], d["images"][0]["height"]
        cat = {c["id"]: c["name"] for c in d["categories"]}
        lines = []
        for a in d["annotations"]:
            name = cat.get(a["category_id"], "")
            cls = "균열" if "균열" in name else ("포트홀" if "홀" in name else None)
            if cls is None: continue
            for s in a["segmentation"]:
                pts = crack_polygon(s, WIDTH_PX) if cls == "균열" else pothole_polygon(s)
                if not pts or len(pts) < 3: stats["bad"] += 1; continue
                if poly_area(pts) > 0.5 * iw * ih: stats["too_big"] += 1; continue
                norm = []
                for x, y in pts:
                    norm += [min(max(x / iw, 0), 1), min(max(y / ih, 0), 1)]
                lines.append(f"{CLASS_MAP[cls]} " + " ".join(f"{c:.6f}" for c in norm))
        src = photo_path_map.get(fn)
        if src is None: stats["no_img"] += 1; continue
        label_out.write_text("\n".join(lines))
        shutil.copy(src, img_out)
        stats["ok"] += 1
    return stats

random.seed(42)
pool = [f for f in crack_map if f not in test_names and f not in set(old_train) and f not in set(old_val)]
extra = random.sample(pool, min(EXTRA_CRACK, len(pool)))
new_train = old_train + extra
print("새 train", len(new_train), "/ val", len(old_val))
with zipfile.ZipFile(TL01_ZIP) as z:
    print("train:", dict(build_split(new_train, "train", z)))
    print("val:", dict(build_split(old_val, "val", z)))
(V2_DIR / "data.yaml").write_text(f"path: {V2_DIR}\ntrain: images/train\nval: images/val\n\nnames:\n  0: 도로균열\n  1: 도로홀\n", encoding="utf-8")

# ===== D. Colab으로 복사 =====
for sub in ["images/train", "images/val", "labels/train", "labels/val"]:
    dst = LOCAL_V2 / sub; dst.mkdir(parents=True, exist_ok=True)
    files = list((V2_DIR / sub).glob("*"))
    for f in files:
        if not (dst / f.name).exists(): shutil.copy(f, dst / f.name)
    print(sub, len(files))
(LOCAL_V2 / "data.yaml").write_text(f"path: {LOCAL_V2}\ntrain: images/train\nval: images/val\n\nnames:\n  0: 도로균열\n  1: 도로홀\n", encoding="utf-8")
DATA = str(LOCAL_V2 / "data.yaml")

# 메모리 정리
for v in ["crack_map", "pothole_map", "filename_to_zipjson", "photo_path_map", "pool", "extra", "new_train"]:
    globals().pop(v, None)
gc.collect()

# ===== E. 1280 학습 (저장본이 있으면 이어하기) =====
from ultralytics import YOLO
name = f"seg_v2_m_1280_{TAG}"
last = RESULT_DIR / name / "weights" / "last.pt"
start = RESULT_DIR / "seg_v2_m_w12_e0" / "weights" / "best.pt"

if last.exists():
    print(f"▶ 이어하기: {name} 에서 계속합니다")
    YOLO(str(last)).train(resume=True)
else:
    print("▶ 새로 시작: 6차 모델에서 1280 + 균열 추가 학습 (25에폭, 약 4~5시간)")
    YOLO(str(start)).train(
        data=DATA, epochs=25, imgsz=1280, batch=4,
        optimizer="AdamW", lr0=0.0005, warmup_epochs=1,
        patience=12, close_mosaic=6, workers=2,
        project=str(RESULT_DIR), name=name,
    )
```
