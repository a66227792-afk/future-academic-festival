```python
import time
import json
import re
import pandas as pd
from pathlib import Path
from google import genai
from google.genai import types
import PIL.Image
from google.colab import userdata

# ====== 설정 ======
API_KEY = userdata.get('GEMINI_API_KEY')
client = genai.Client(api_key=API_KEY)

PHOTO_FOLDER = Path("/content/drive/MyDrive/미래학술제/New_Sample (1)/원천데이터/TS01_지자체도로정비AI데이터_CASE1(도로)/지자체도로정비AI데이터_CASE1(도로)_WET_RAIN_01")
LABEL_FOLDER = Path("/content/drive/MyDrive/미래학술제/New_Sample (1)/라벨링데이터/TL01_지자체도로정비AI데이터_CASE1(도로)/지자체도로정비AI데이터_CASE1(도로)_WET_RAIN_01")

N = 20

PROMPT = """이 사진 속 도로 파손을 분석해줘. 아래 형식으로만 정확히 답해줘 (다른 말 붙이지 말고):
유형: 균열 또는 포트홀 (둘 중 하나만)
심각도: 경미 또는 보통 또는 심각 또는 긴급 (하나만)"""

def get_ground_truth(json_path):
    with open(json_path, "r", encoding="utf-8") as f:
        data = json.load(f)
    cat_map = {c["id"]: c["name"] for c in data["categories"]}
    if not data["annotations"]:
        return None
    cat_id = data["annotations"][0]["category_id"]
    name = cat_map.get(cat_id, "")
    if "균열" in name:
        return "균열"
    elif "홀" in name:
        return "포트홀"
    return None

def ask_gemini(image_path):
    image = PIL.Image.open(image_path)
    wait = 2  # 재시도 대기시간, 실패할 때마다 2배씩 늘어남 (2→4→8)
    for attempt in range(4):
        try:
            response = client.models.generate_content(
                model="gemini-flash-lite-latest",   # 더 가벼운 모델로 변경
                contents=[image, PROMPT]
            )
            return response.text
        except Exception:
            time.sleep(wait)
            wait *= 2
    return "실패"

# ====== 실행 ======
results = []
import random

all_photos = sorted(PHOTO_FOLDER.glob("*.jpg"))
random.seed(42)
random.shuffle(all_photos)
photos = all_photos[:N]

for i, photo_path in enumerate(photos):
    print(f"처리 중... {i+1}/{len(photos)} : {photo_path.name}")

    json_path = LABEL_FOLDER / (photo_path.stem + ".json")
    if not json_path.exists():
        print("  -> 라벨 없음, 건너뜀")
        continue

    ground_truth = get_ground_truth(json_path)
    ai_answer = ask_gemini(photo_path)

    match = re.search(r"유형:\s*(균열|포트홀)", ai_answer)
    ai_type = match.group(1) if match else "파싱실패"

    severity_match = re.search(r"심각도:\s*(경미|보통|심각|긴급)", ai_answer)
    ai_severity = severity_match.group(1) if severity_match else "파싱실패"

    is_correct = "O" if ai_type == ground_truth else "X"

    results.append({
        "사진명": photo_path.name,
        "정답유형": ground_truth,
        "AI답유형": ai_type,
        "맞았는지": is_correct,
        "AI심각도": ai_severity,
        "전체답변": ai_answer,
    })

    time.sleep(3)  # 4초 -> 3초로 살짝 단축

df = pd.DataFrame(results)
print("\n=== 전체 결과 ===")
print(df[["사진명", "정답유형", "AI답유형", "맞았는지", "AI심각도"]])

accuracy = (df["맞았는지"] == "O").sum() / len(df) * 100
print(f"\n정확도: {accuracy:.1f}%")

save_path = "/content/drive/MyDrive/미래학술제/조민서/gemini_테스트결과.xlsx"
df.to_excel(save_path, index=False)
print(f"\n저장 완료: {save_path}")
```

```python
import random

# 포트홀 사진들의 라벨(json) 경로를 먼저 다 찾기
pothole_jsons = []
crack_jsons = []

for jf in all_json:
    with open(jf, "r", encoding="utf-8") as f:
        data = json.load(f)
    cat_map = {c["id"]: c["name"] for c in data["categories"]}
    if data["annotations"]:
        cat_id = data["annotations"][0]["category_id"]
        name = cat_map.get(cat_id)
        if name == "도로(홀)":
            pothole_jsons.append(jf)
        elif name == "도로균열":
            crack_jsons.append(jf)

print(f"포트홀 라벨: {len(pothole_jsons)}개, 균열 라벨: {len(crack_jsons)}개")

# 포트홀 15장 전부 + 균열 5장 무작위로 = 20장
random.seed(42)
selected_crack = random.sample(crack_jsons, 15)   # 5 → 15로 변경
selected_labels = pothole_jsons + selected_crack

print(f"\n최종 테스트셋: 포트홀 {len(pothole_jsons)}장 + 균열 {len(selected_crack)}장 = {len(selected_labels)}장")
```

```python
def ask_claude(image_path):
    base64_image = encode_image(image_path)
    wait = 2
    for attempt in range(4):
        try:
            response = client_qwen.chat.completions.create(
                model="anthropic/claude-haiku-4.5",
                max_tokens=100,
                messages=[
                    {
                        "role": "user",
                        "content": [
                            {"type": "text", "text": PROMPT},
                            {"type": "image_url", "image_url": {"url": f"data:image/jpeg;base64,{base64_image}"}}
                        ]
                    }
                ]
            )
            return response.choices[0].message.content
        except Exception:
            time.sleep(wait)
            wait *= 2
    return "실패"
```

```python
import concurrent.futures

def process_one(json_path):
    photo_path = json_to_photo_path(json_path)
    if not photo_path.exists():
        return None

    ground_truth = get_ground_truth_from_json(json_path)
    ai_answer = ask_claude(photo_path)

    match = re.search(r"유형:\s*(균열|포트홀)", ai_answer)
    ai_type = match.group(1) if match else "파싱실패"

    severity_match = re.search(r"심각도:\s*(경미|보통|심각|긴급)", ai_answer)
    ai_severity = severity_match.group(1) if severity_match else "파싱실패"

    is_correct = "O" if ai_type == ground_truth else "X"

    return {
        "사진명": photo_path.name,
        "정답유형": ground_truth,
        "AI답유형": ai_type,
        "맞았는지": is_correct,
        "AI심각도": ai_severity,
    }

# ====== 실행 (동시에 5개씩 처리) ======
results_claude = []

with concurrent.futures.ThreadPoolExecutor(max_workers=5) as executor:
    futures = {executor.submit(process_one, jp): jp for jp in selected_labels}
    for i, future in enumerate(concurrent.futures.as_completed(futures)):
        result = future.result()
        if result:
            results_claude.append(result)
            print(f"완료 {len(results_claude)}/{len(selected_labels)} : {result['사진명']}")

df_claude = pd.DataFrame(results_claude)
print("\n=== Claude 전체 결과 ===")
print(df_claude)

accuracy_overall = (df_claude["맞았는지"] == "O").sum() / len(df_claude) * 100
print(f"\n전체 정확도: {accuracy_overall:.1f}%")

for gt in ["균열", "포트홀"]:
    subset = df_claude[df_claude["정답유형"] == gt]
    if len(subset) > 0:
        acc = (subset["맞았는지"] == "O").sum() / len(subset) * 100
        print(f"{gt} 정확도: {acc:.1f}% ({len(subset)}장 중)")

save_path = "/content/drive/MyDrive/미래학술제/조민서/claude_테스트결과.xlsx"
df_claude.to_excel(save_path, index=False)
print(f"\n저장 완료: {save_path}")
```

```python
import pandas as pd

# ====== 1. 5개 결과 파일 불러오기 ======
base_path = "/content/drive/MyDrive/미래학술제/조민서/"

df_gemini = pd.read_excel(base_path + "gemini_테스트결과.xlsx")
df_qwen = pd.read_excel(base_path + "qwen_테스트결과.xlsx")
df_claude = pd.read_excel(base_path + "claude_테스트결과.xlsx")
df_gpt = pd.read_excel(base_path + "gpt4o_테스트결과.xlsx")
df_clova = pd.read_excel(base_path + "clova_테스트결과.xlsx")

models = {
    "Gemini": df_gemini,
    "Qwen": df_qwen,
    "Claude": df_claude,
    "GPT-4o": df_gpt,
    "HyperCLOVA X": df_clova,
}

# ====== 2. 사진별로 5개 모델 답을 나란히 놓기 ======
merged = df_gemini[["사진명", "정답유형"]].copy()

for name, df in models.items():
    temp = df.set_index("사진명")
    merged[f"{name}_답"] = merged["사진명"].map(temp["AI답유형"])
    merged[f"{name}_맞음"] = merged["사진명"].map(temp["맞았는지"])

print("=== 사진별 5개 모델 비교표 (일부) ===")
print(merged.head(10))

# ====== 3. 모델별 요약 통계표 ======
summary_rows = []
for name, df in models.items():
    overall = (df["맞았는지"] == "O").sum() / len(df) * 100
    crack = df[df["정답유형"] == "균열"]
    pothole = df[df["정답유형"] == "포트홀"]
    crack_acc = (crack["맞았는지"] == "O").sum() / len(crack) * 100 if len(crack) > 0 else None
    pothole_acc = (pothole["맞았는지"] == "O").sum() / len(pothole) * 100 if len(pothole) > 0 else None

    summary_rows.append({
        "모델": name,
        "전체 정확도(%)": round(overall, 1),
        "균열 정확도(%)": round(crack_acc, 1),
        "포트홀 정확도(%)": round(pothole_acc, 1),
    })

summary_df = pd.DataFrame(summary_rows).sort_values("전체 정확도(%)", ascending=False)
print("\n=== 모델별 요약 비교표 ===")
print(summary_df)

# ====== 4. 하나의 엑셀 파일로 저장 (시트 3개로 구분) ======
save_path = base_path + "5개모델_종합비교.xlsx"

with pd.ExcelWriter(save_path, engine="openpyxl") as writer:
    summary_df.to_excel(writer, sheet_name="요약비교", index=False)
    merged.to_excel(writer, sheet_name="사진별상세비교", index=False)
    for name, df in models.items():
        df.to_excel(writer, sheet_name=name, index=False)

print(f"\n저장 완료: {save_path}")
```

```python
import time
import re
import json
import pandas as pd
from pathlib import Path
from google import genai
from google.colab import userdata
import PIL.Image

API_KEY = userdata.get('GEMINI_API_KEY')
client = genai.Client(api_key=API_KEY)

PROMPT = """이 사진 속 도로 파손을 분석해줘. 아래 형식으로만 정확히 답해줘 (다른 말 붙이지 말고):

유형: 균열 또는 포트홀 (둘 중 하나만)

심각도: 아래 기준을 참고해서 판단해줘
- 경미: 표면에 실금 수준이거나 아주 작은 파임, 즉각적인 위험 없음
- 보통: 육안으로 뚜렷이 보이는 파손, 주의가 필요한 수준
- 심각: 파손 범위가 넓거나 깊어 통행에 지장을 줄 수 있는 수준
- 긴급: 깊은 구멍, 넓은 함몰, 도로 일부 붕괴 등 사고 위험이 임박한 수준

위 4단계 중 하나만 답해줘."""

def ask_gemini(image_path):
    image = PIL.Image.open(image_path)
    wait = 2
    for attempt in range(4):
        try:
            response = client.models.generate_content(
                model="gemini-flash-lite-latest",
                contents=[image, PROMPT]
            )
            return response.text
        except Exception:
            time.sleep(wait)
            wait *= 2
    return "실패"

def get_ground_truth(json_path):
    with open(json_path, "r", encoding="utf-8") as f:
        data = json.load(f)
    cat_map = {c["id"]: c["name"] for c in data["categories"]}
    cat_id = data["annotations"][0]["category_id"]
    name = cat_map.get(cat_id, "")
    return "균열" if "균열" in name else "포트홀"

# ====== 실행 ======
test_photo_folder = Path("/content/drive/MyDrive/미래학술제/최종테스트셋/최종_원천")
test_label_folder = Path("/content/drive/MyDrive/미래학술제/최종테스트셋/최종_라벨")

results = []
label_files = sorted(test_label_folder.glob("*.json"))
total = len(label_files)
start_time = time.time()

for i, json_path in enumerate(label_files):
    photo_path = test_photo_folder / (json_path.stem + ".jpg")

    if not photo_path.exists():
        continue

    ground_truth = get_ground_truth(json_path)
    ai_answer = ask_gemini(photo_path)

    match = re.search(r"유형[:\*\s]*\**\s*(균열|포트홀)", ai_answer)
    ai_type = match.group(1) if match else ""

    severity_match = re.search(r"(경미|보통|심각|긴급)", ai_answer)
    ai_severity = severity_match.group(1) if severity_match else ""

    results.append({
        "사진 번호": i + 1,
        "파일명": photo_path.name,
        "정답 유형(라벨 원본)": ground_truth,
        "AI 판단 유형": ai_type,
        "AI 판단 심각도": ai_severity,
        "사용 모델": "Gemini",
    })

    done = i + 1
    if done % 10 == 0 or done == total:
        elapsed = time.time() - start_time
        percent = done / total * 100
        speed = elapsed / done
        remaining = (total - done) * speed

        print(f"[{done}/{total}] {percent:.1f}% 완료 | "
              f"경과 {int(elapsed//60)}분{int(elapsed%60)}초 | "
              f"예상 남은 시간 {int(remaining//60)}분{int(remaining%60)}초")

    time.sleep(2)

df_gemini = pd.DataFrame(results)
save_path = "/content/drive/MyDrive/미래학술제/조민서/gemini_300장_결과_v2.xlsx"
df_gemini.to_excel(save_path, index=False)

total_elapsed = time.time() - start_time
print(f"\n=== 전체 완료! ===")
print(f"총 소요 시간: {int(total_elapsed//60)}분 {int(total_elapsed%60)}초")
print(f"저장 완료: {save_path}")
print(f"전체 {len(df_gemini)}건 처리됨")

# ====== 결과 요약 (바로 확인용) ======
from collections import Counter
correct = (df_gemini["정답 유형(라벨 원본)"] == df_gemini["AI 판단 유형"]).sum()
print(f"\n전체 정확도: {correct/len(df_gemini)*100:.1f}%")

for gt in ["균열", "포트홀"]:
    subset = df_gemini[df_gemini["정답 유형(라벨 원본)"] == gt]
    acc = (subset["정답 유형(라벨 원본)"] == subset["AI 판단 유형"]).sum() / len(subset) * 100
    print(f"{gt} 정확도: {acc:.1f}%")

print("\n심각도 분포:", dict(Counter(df_gemini["AI 판단 심각도"])))
```

```python
import time
import json
import pandas as pd
import concurrent.futures
from pathlib import Path
from google import genai
from google.colab import userdata
import PIL.Image

API_KEY = userdata.get('GEMINI_API_KEY')
client = genai.Client(api_key=API_KEY)

PROMPT = """이 사진 속 도로 파손을 분석해줘. 아래 형식으로만 정확히 답해줘 (다른 말 붙이지 말고):

유형: 균열 또는 포트홀 (둘 중 하나만)

심각도: 아래 기준을 참고해서 판단해줘
- 경미: 표면에 실금 수준이거나 아주 작은 파임, 즉각적인 위험 없음
- 보통: 육안으로 뚜렷이 보이는 파손, 주의가 필요한 수준
- 심각: 파손 범위가 넓거나 깊어 통행에 지장을 줄 수 있는 수준
- 긴급: 깊은 구멍, 넓은 함몰, 도로 일부 붕괴 등 사고 위험이 임박한 수준

반드시 아래 형식 그대로만 답하고, 기준 설명은 다시 쓰지 마:
유형: [답]
심각도: [답]"""

def ask_gemini(image_path):
    image = PIL.Image.open(image_path)
    wait = 2
    for attempt in range(4):
        try:
            response = client.models.generate_content(
                model="gemini-flash-lite-latest",
                contents=[image, PROMPT]
            )
            return response.text
        except Exception:
            time.sleep(wait)
            wait *= 2
    return "실패"

def get_ground_truth(json_path):
    with open(json_path, "r", encoding="utf-8") as f:
        data = json.load(f)
    cat_map = {c["id"]: c["name"] for c in data["categories"]}
    cat_id = data["annotations"][0]["category_id"]
    name = cat_map.get(cat_id, "")
    return "균열" if "균열" in name else "포트홀"

def process_one(args):
    i, json_path = args
    photo_path = test_photo_folder / (json_path.stem + ".jpg")
    if not photo_path.exists():
        return None
    ground_truth = get_ground_truth(json_path)
    ai_answer = ask_gemini(photo_path)
    return {
        "사진 번호": i + 1,
        "파일명": photo_path.name,
        "정답 유형(라벨 원본)": ground_truth,
        "AI 원본 답변": ai_answer,
        "사용 모델": "Gemini",
    }

# ====== 실행: 3개씩 동시 처리 (너무 공격적이지 않게) ======
test_photo_folder = Path("/content/drive/MyDrive/미래학술제/최종테스트셋/최종_원천")
test_label_folder = Path("/content/drive/MyDrive/미래학술제/최종테스트셋/최종_라벨")

label_files = sorted(test_label_folder.glob("*.json"))
total = len(label_files)
start_time = time.time()

raw_results = []
with concurrent.futures.ThreadPoolExecutor(max_workers=3) as executor:
    futures = {executor.submit(process_one, (i, jp)): jp for i, jp in enumerate(label_files)}
    done = 0
    for future in concurrent.futures.as_completed(futures):
        result = future.result()
        if result:
            raw_results.append(result)
        done += 1
        if done % 10 == 0 or done == total:
            elapsed = time.time() - start_time
            remaining = (total - done) * (elapsed / done)
            print(f"[{done}/{total}] {done/total*100:.1f}% | "
                  f"경과 {int(elapsed//60)}분{int(elapsed%60)}초 | "
                  f"남은 {int(remaining//60)}분{int(remaining%60)}초")

df_raw = pd.DataFrame(raw_results)
raw_save_path = "/content/drive/MyDrive/미래학술제/조민서/gemini_300장_원본답변_v3.xlsx"
df_raw.to_excel(raw_save_path, index=False)

total_elapsed = time.time() - start_time
print(f"\n완료! 총 {int(total_elapsed//60)}분{int(total_elapsed%60)}초")
print(f"저장 완료: {raw_save_path}")
```

```python
def ask_gemini_safe(image_path):
    image = PIL.Image.open(image_path)
    wait = 3
    for attempt in range(4):
        try:
            response = client.models.generate_content(
                model="gemini-flash-lite-latest",
                contents=[image, PROMPT]
            )
            return response.text
        except Exception:
            time.sleep(wait)
            wait *= 2
    return "실패"

def get_ground_truth(json_path):
    with open(json_path, "r", encoding="utf-8") as f:
        data = json.load(f)
    cat_map = {c["id"]: c["name"] for c in data["categories"]}
    cat_id = data["annotations"][0]["category_id"]
    name = cat_map.get(cat_id, "")
    return "균열" if "균열" in name else "포트홀"

label_files = sorted(test_label_folder.glob("*.json"))
total = len(label_files)
start_time = time.time()

raw_results = []
fail_count = 0

for i, json_path in enumerate(label_files):
    photo_path = test_photo_folder / (json_path.stem + ".jpg")
    if not photo_path.exists():
        continue

    ground_truth = get_ground_truth(json_path)
    ai_answer = ask_gemini_safe(photo_path)

    if ai_answer == "실패":
        fail_count += 1

    raw_results.append({
        "사진 번호": i + 1,
        "파일명": photo_path.name,
        "정답 유형(라벨 원본)": ground_truth,
        "AI 원본 답변": ai_answer,
        "사용 모델": "Gemini",
    })

    done = i + 1
    if done % 10 == 0 or done == total:
        elapsed = time.time() - start_time
        remaining = (total - done) * (elapsed / done)
        print(f"[{done}/{total}] {done/total*100:.1f}% | "
              f"경과 {int(elapsed//60)}분{int(elapsed%60)}초 | "
              f"남은 {int(remaining//60)}분{int(remaining%60)}초 | "
              f"실패 {fail_count}건")

        if fail_count >= 15:
            print("\n⚠️ 실패 15건 이상 - 자동 중단")
            break

    time.sleep(3)

df_raw = pd.DataFrame(raw_results)
raw_save_path = "/content/drive/MyDrive/미래학술제/조민서/gemini_300장_최종.xlsx"
df_raw.to_excel(raw_save_path, index=False)

print(f"\n완료! {len(df_raw)}건 처리 (실패 {fail_count}건)")
print(f"저장: {raw_save_path}")
```

```python
from openai import OpenAI
from google.colab import userdata
import base64, time
import pandas as pd

CLOVA_API_KEY = userdata.get('CLOVA_API_KEY')
client_clova = OpenAI(
    api_key=CLOVA_API_KEY,
    base_url="https://clovastudio.stream.ntruss.com/v1/openai"
)

def encode_image(image_path):
    with open(image_path, "rb") as f:
        return base64.b64encode(f.read()).decode("utf-8")

def ask_clova_safe(image_path):
    base64_image = encode_image(image_path)
    wait = 3
    for attempt in range(4):
        try:
            response = client_clova.chat.completions.create(
                model="HCX-005",
                max_tokens=100,
                messages=[{
                    "role": "user",
                    "content": [
                        {"type": "text", "text": PROMPT},
                        {"type": "image_url", "image_url": {"url": f"data:image/jpeg;base64,{base64_image}"}}
                    ]
                }]
            )
            return response.choices[0].message.content
        except Exception:
            time.sleep(wait)
            wait *= 2
    return "실패"

test_photo = list(test_photo_folder.glob("*.jpg"))[0]
try:
    print("✅ 성공:", ask_clova_safe(test_photo))
except Exception as e:
    print("❌ 실패:", e)
```

## 1. 평가 개요

- **평가 대상**: Gemini(Flash Lite), HyperCLOVA X(HCX-005)
- **테스트셋**: 도로 균열 150장 + 포트홀 150장, 총 300장 (무작위 추출, seed=42 고정)
- **평가 방식**: 동일한 프롬프트로 유형(균열/포트홀)과 심각도(경미/보통/심각/긴급)를 판단하게 하여, 라벨 원본과 비교
- **API 실행**: 두 모델 모두 실패 0건, 300장 전부 정상 처리 완료

## 2. Gemini 결과 분석

### 정확도

| 구분 | 정확도 |
|---|---|
| 전체 | 71.3% |
| 균열 | **94.7%** |
| 포트홀 | 48.0% |

### 심각도 분포

경미 128건 / 보통 167건 / 심각 5건 / 긴급 0건

### 해석

- **균열 인식은 매우 우수함**(94.7%) — 거의 완벽에 가까운 수준
- **포트홀 인식은 절반 수준(48.0%)으로 저조** — 균열과 포트홀을 혼동하는 경향이 뚜렷함
- 심각도 판단이 "경미"와 "보통"에 극단적으로 쏠려있고, "심각"은 5건뿐, "긴급"은 전무함 → **위험도를 보수적으로 낮게 판단하는 경향**
- 전반적으로 두 유형 간 인식 능력 격차가 매우 크다는 것이 확인됨

## 3. HyperCLOVA X 결과 분석

### 정확도

| 구분 | 정확도 |
|---|---|
| 전체 | **76.0%** |
| 균열 | 78.0% |
| 포트홀 | **74.0%** |

### 심각도 분포

경미 141건 / 보통 105건 / 심각 49건 / 긴급 0건

### "해당 없음" 응답에 대한 참고사항

300장 중 5건(1.7%)에서 유형·심각도가 파싱되지 않았는데, 원본 답변을 직접 확인한 결과 이는 정규식 파싱 오류가 아니라 **HyperCLOVA X가 "유형: 해당 없음, 심각도: 해당 없음"이라고 명시적으로 답한 경우**였음이 확인됨. 5건 중 4건은 정답 라벨이 균열, 1건은 포트홀이었다. 이 5건은 정답을 맞히지 못한 것이므로 정확도 계산에서 오답(X)으로 정확히 반영되었으며, 별도의 재계산은 필요하지 않음.

### 해석

- **균열과 포트홀 정확도가 78.0% / 74.0%로 균형 잡혀 있음** — Gemini처럼 한쪽으로 치우치지 않음
- **포트홀 인식이 Gemini 대비 압도적으로 우수함**(74.0% vs 48.0%, 26%p 차이)
- 심각도 판단에서 "심각" 응답이 49건으로, Gemini(5건)보다 훨씬 적극적으로 판단함 → 더 세분화된 위험도 평가가 가능함을 시사
- "긴급" 응답은 Gemini와 동일하게 0건으로, 이는 특정 모델의 한계가 아니라 **프롬프트 설계 또는 데이터 자체의 특성**일 가능성을 시사함 (후속 논의 필요)
- 소수(1.7%)의 "판단 불가" 응답이 존재하나, 전체 결과 해석에 미치는 영향은 미미함

## 4. Gemini vs HyperCLOVA X 종합 비교

| 지표 | Gemini | HyperCLOVA X | 우세 |
|---|---|---|---|
| 전체 정확도 | 71.3% | 76.0% | HyperCLOVA X |
| 균열 정확도 | 94.7% | 78.0% | Gemini |
| 포트홀 정확도 | 48.0% | 74.0% | **HyperCLOVA X (큰 격차)** |
| 심각도 판단 다양성 | 3단계 사용(심각 5건) | 3단계 사용(심각 49건) | HyperCLOVA X |
| "판단 불가" 응답 | 0건 | 5건(1.7%) | Gemini |

**핵심 발견**: HyperCLOVA X가 전체 정확도와 포트홀 인식에서 뚜렷하게 우수하며, 특히 포트홀 인식 격차(26%p)가 결정적임. Gemini는 균열 인식에 한해서는 여전히 강점을 보이며, 판단을 회피하지 않고 항상 답을 내놓는 경향을 보임.

## 5. 메인/보조 모델 재조정

### 기존 선정 (30장 예비 실험 기준)

- **메인**: Gemini — 이유: 30장 예비 테스트에서 정확도가 HyperCLOVA X와 동률(83.3%)이었으며, 무료라는 비용 이점이 결정적 근거였음
- **보조**: HyperCLOVA X — 국내 특화 가능성 고려한 비교 대상

### 재조정 결정: 메인 ↔ 보조 교체

- **새로운 메인**: **HyperCLOVA X**
- **새로운 보조**: **Gemini**

### 재조정 근거

1. **정확도 격차가 명확해짐**: 300장 공식 평가에서 HyperCLOVA X가 전체 정확도(76.0% vs 71.3%)와 포트홀 인식(74.0% vs 48.0%)에서 뚜렷한 우위를 보임. 30장 예비 실험의 "동률"은 표본이 작아 나타난 우연이었을 가능성이 크며, 300장 표본에서 실제 성능 차이가 명확히 드러남. 파싱 실패 5건을 원본까지 확인해 오답으로 정확히 반영한 뒤에도 이 우위는 그대로 유지됨
2. **비용 기준의 무게가 감소함**: 본 연구는 상용 서비스가 아닌 학술제 프로토타입으로, 300장 1회 평가와 데모 시연 수준에서만 API를 사용하므로 유료 비용 부담이 사실상 무시할 만한 수준임. 애초에 Gemini를 메인으로 선택하게 했던 핵심 근거(비용)의 실효성이 약화됨
3. **연구 스토리와도 부합함**: HyperCLOVA X는 국내 기업이 개발한 모델로, 국내 도로 데이터에서 더 높은 정확도를 보인 것은 "국내 특화 모델이 국내 실환경 데이터에 더 적합할 수 있다"는 흥미로운 시사점을 제공하며 발표 자료의 논리를 보강함

```python
import openpyxl
import pandas as pd
import re
from pathlib import Path

yoonsu_excel_path = "/content/drive/MyDrive/미래학술제/신윤수/위험요소_채점&기록시트.xlsx"

df_gemini = pd.read_excel("/content/drive/MyDrive/미래학술제/조민서/gemini_300장_최종.xlsx")
df_clova = pd.read_excel("/content/drive/MyDrive/미래학술제/조민서/clova_300장_최종.xlsx")

def parse_type(text):
    m = re.search(r"유형[:\s]*(균열|포트홀)", str(text))
    return m.group(1) if m else ""

def parse_severity(text):
    m = re.search(r"심각도[:\s]*(경미|보통|심각|긴급)", str(text))
    return m.group(1) if m else ""

for df in [df_gemini, df_clova]:
    df["AI 판단 유형"] = df["AI 원본 답변"].apply(parse_type)
    df["AI 판단 심각도"] = df["AI 원본 답변"].apply(parse_severity)

df_clova = df_clova.reset_index(drop=True)
df_clova["최종사진번호"] = range(1, len(df_clova) + 1)

df_gemini = df_gemini.reset_index(drop=True)
df_gemini["최종사진번호"] = range(len(df_clova) + 1, len(df_clova) + len(df_gemini) + 1)

df_all = pd.concat([df_clova, df_gemini], ignore_index=True)

wb = openpyxl.load_workbook(yoonsu_excel_path)

# ====== 병합된 셀 먼저 풀어주기 (에러 방지) ======
def unmerge_all(ws):
    merged_ranges = list(ws.merged_cells.ranges)
    for merged_range in merged_ranges:
        ws.unmerge_cells(str(merged_range))

ws1 = wb["AI 채점용"]
ws2 = wb["심각도 신뢰성"]

unmerge_all(ws1)
unmerge_all(ws2)

# ---- 시트1: AI 채점용 ----
for i, row in df_all.iterrows():
    excel_row = i + 3
    ws1.cell(row=excel_row, column=1, value=row["최종사진번호"])
    ws1.cell(row=excel_row, column=2, value=row["파일명"])
    ws1.cell(row=excel_row, column=3, value=row["정답 유형(라벨 원본)"])
    ws1.cell(row=excel_row, column=4, value=row["AI 판단 유형"])
    ws1.cell(row=excel_row, column=5, value=row["사용 모델"])

# ---- 시트2: 심각도 신뢰성 ----
for i, row in df_all.iterrows():
    excel_row = i + 3
    ws2.cell(row=excel_row, column=1, value=row["최종사진번호"])
    ws2.cell(row=excel_row, column=2, value=row["파일명"])
    ws2.cell(row=excel_row, column=3, value=row["AI 판단 심각도"])
    ws2.cell(row=excel_row, column=4, value=row["사용 모델"])

# ====== 저장 ======
save_path_yoonsu = "/content/drive/MyDrive/미래학술제/신윤수/위험요소_채점&기록시트_결과입력완료.xlsx"
save_path_minseo = "/content/drive/MyDrive/미래학술제/조민서/위험요소_채점&기록시트_결과입력완료.xlsx"

wb.save(save_path_yoonsu)
wb.save(save_path_minseo)

print(f"저장 완료 (신윤수 폴더): {save_path_yoonsu}")
print(f"저장 완료 (조민서 폴더): {save_path_minseo}")
print(f"총 {len(df_all)}행 입력됨 (HyperCLOVA X 1~{len(df_clova)}, Gemini {len(df_clova)+1}~{len(df_all)})")
```


## 문제 상황

파싱 코드가 "균열|포트홀" 중 하나를 찾는 방식으로 작성되어 있어, "해당 없음"이라는 응답은
정규식에 매칭되지 않고 **빈 값(공란)**으로 처리되어 엑셀 "AI 채점용" 시트 D열과
"심각도 신뢰성" 시트 C열에 공란으로 입력되었음.

빈 값으로 둘 경우 발생하는 문제:

- F열 자동 계산 수식 `=IF(AND(C≠"",D≠""),IF(C=D,"O","X"),"")`이 D열 공란 조건에 걸려
  해당 5건이 O도 X도 아닌 **빈칸**으로 남게 됨
- 이 5건이 전체 정확도(%) 계산의 분모(전체 300장)에 포함되는지 불분명해짐
- 데이터만 봤을 때 "AI가 판단을 못 한 것"인지 "입력이 누락된 것"인지 구분이 안 됨
- 발표 시 "왜 이 칸들이 비어있나요?"라는 질문에 명확히 답하기 어려움

## 처리 방침: 빈칸 대신 "해당없음" 텍스트로 명시

| 항목 | 빈칸으로 둘 경우 | "해당없음"으로 채울 경우 |
|---|---|---|
| F열(O/X) 계산 | 빈칸 (판정 불가) | **자동으로 "X"(오답) 처리** |
| 정확도 계산 반영 | 불명확 | 명확하게 오답으로 반영됨 |
| 데이터 해석 | AI 미판단인지 입력 누락인지 구분 불가 | AI가 명시적으로 판단을 포기했음을 그대로 기록 |
| 보고서와의 일치성 | 데이터와 서술이 어긋날 수 있음 | "5건은 AI가 판단 불가를 표명했다"는 서술과 정확히 일치 |

## 결론

AI가 "판단을 내리지 못한 것"도 정직한 성능 결과의 일부이며, 이를 숨기지 않고
데이터에 그대로 반영하는 것이 학술적으로 더 정확한 태도임. 이에 따라 "AI 채점용" 시트
D열과 "심각도 신뢰성" 시트 C열의 해당 5개 행에 빈칸 대신 **"해당없음"**을 명시적으로 입력하였고,
이 값이 정답(균열/포트홀, 경미~긴급)과 자동으로 불일치 처리되어 오답(X)으로 정확히
집계되도록 하였다.

```python
import openpyxl
from pathlib import Path

file_path = "/content/drive/MyDrive/미래학술제/조민서/위험요소_채점&기록시트_결과입력완료.xlsx"

wb = openpyxl.load_workbook(file_path)
ws1 = wb["AI 채점용"]
ws2 = wb["심각도 신뢰성"]

# ====== 수정 대상: 사진번호 64, 119, 126, 160, 196 ======
target_photo_numbers = [64, 119, 126, 160, 196]

# ---- AI 채점용 시트: D열(AI 판단 유형) 채우기 ----
fixed_count_1 = 0
for r in range(3, ws1.max_row + 1):
    photo_num = ws1.cell(row=r, column=1).value
    if photo_num in target_photo_numbers:
        ws1.cell(row=r, column=4, value="해당없음")  # D열
        fixed_count_1 += 1

# ---- 심각도 신뢰성 시트: C열(AI 판단 심각도)도 같이 확인 ----
fixed_count_2 = 0
for r in range(3, ws2.max_row + 1):
    photo_num = ws2.cell(row=r, column=1).value
    if photo_num in target_photo_numbers:
        current_val = ws2.cell(row=r, column=3).value  # C열
        if current_val is None or current_val == "":
            ws2.cell(row=r, column=3, value="해당없음")
            fixed_count_2 += 1

# ====== 저장 ======
wb.save(file_path)

print(f"AI 채점용 시트: {fixed_count_1}건 수정 완료")
print(f"심각도 신뢰성 시트: {fixed_count_2}건 수정 완료")
print(f"저장 완료: {file_path}")
```
